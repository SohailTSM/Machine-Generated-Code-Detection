```python
# Cell 2 — Secrets + copy code and test files
import os, shutil
from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()
os.environ["HF_READ_TOKEN"] = user_secrets.get_secret("HF_READ_TOKEN")
os.environ["HF_WRITE_TOKEN"] = user_secrets.get_secret("HF_WRITE_TOKEN")
os.environ["WANDB_API_KEY"] = user_secrets.get_secret("WANDB_API_KEY")

CODE_DS = "/kaggle/input/datasets/dhruvrajput1050/graph-transformer/INLP project GNN"
TEST_DS = "/kaggle/input/datasets/dhruvrajput1050/final-tests/INLP project GNN/final_test_files"

# Copy source code
for item in ["run.py", "generate_submission.py", "requirements.txt", "src"]:
    src = os.path.join(CODE_DS, item)
    dst = os.path.join("/kaggle/working", item)
    if os.path.isdir(src):
        shutil.copytree(src, dst, dirs_exist_ok=True)
    else:
        shutil.copy2(src, dst)

# Copy test files
shutil.copytree(TEST_DS, "/kaggle/working/final_test_files", dirs_exist_ok=True)

print("✓ Code + test files copied")
print("  Artifacts will be downloaded from HuggingFace automatically")
```
